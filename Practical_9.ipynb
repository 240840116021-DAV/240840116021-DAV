Step 2: Import libraries

```python
# Import required libraries
import pandas as pd
import plotly.express as px
# Import Dash components
from dash import Dash, dcc, html, Input, Output
```

Step 3: Create a different dataset

```python
# Create electricity consumption dataset
data = {
    "Month": [
        "Jan", "Feb", "Mar", "Apr",
        "May", "Jun", "Jul", "Aug",
        "Sep", "Oct", "Nov", "Dec"
    ],

    "City": [
        "Surat", "Mumbai", "Pune", "Nashik",
        "Surat", "Mumbai", "Pune", "Nashik",
        "Surat", "Mumbai", "Pune", "Nashik"
    ],

    "Appliance": [
        "AC", "Fan", "TV", "Refrigerator",
        "Washing Machine", "AC", "Fan", "TV",
        "Refrigerator", "Washing Machine", "AC", "Fan"
    ],

    "Units_Consumed": [
        250, 120, 150, 180,
        130, 280, 140, 160,
        190, 145, 300, 135
    ],

    "Electricity_Bill": [
        1800, 900, 1100, 1300,
        950, 2000, 1000, 1150,
        1400, 1050, 2200, 980
    ]
}
# Convert data into DataFrame
df = pd.DataFrame(data)
# Display dataset
print(df)
```

Step 4: Display dataset information

```python
# Display first five records
print(df.head())

# Display dataset information
print(df.info())

# Display statistical information
print(df.describe())

# Check missing values
print(df.isnull().sum())
```

Step 5: Create dashboard layout

```python
# Create Dash application
app = Dash(__name__)

# Design dashboard layout
app.layout = html.Div([

   # Dashboard title
    html.H1(
        "Electricity Consumption Dashboard",
        style={
            "textAlign": "center",
            "color": "darkblue"
        }
    ),
   # City selection
    html.H3("Select City"),

     # Create dropdown menu
    dcc.Dropdown(
        id="city",
        options=[
            {"label": "All Cities", "value": "All"}
        ] + [
            {"label": city, "value": city}
            for city in df["City"].unique()
        ],
        value="All",
        clearable=False
    ),
# Display summary
    html.Div(id="summary"),
# Display charts
    dcc.Graph(id="units_chart"),

    dcc.Graph(id="bill_chart"),

    dcc.Graph(id="appliance_chart")

], style={"padding": "20px"})
```

Step 6: Add interactive functionality

```python
# Update dashboard based on selected city
@app.callback(
    Output("summary", "children"),
    Output("units_chart", "figure"),
    Output("bill_chart", "figure"),
    Output("appliance_chart", "figure"),
    Input("city", "value")
)
# Define dashboard function
def update_dashboard(city):

    if city == "All":
        filtered = df
    else:
        filtered = df[df["City"] == city]

    # Calculate total units and bill
    total_units = filtered["Units_Consumed"].sum()
    total_bill = filtered["Electricity_Bill"].sum()

    # Display summary
    summary = html.Div([
        html.H3(f"Total Units Consumed: {total_units} kWh"),
        html.H3(f"Total Electricity Bill: ₹{total_bill}"),
        html.H3(f"Total Records: {len(filtered)}")
    ])

    # Line chart
    units_fig = px.line(
        filtered,
        x="Month",
        y="Units_Consumed",
        markers=True,
        title="Monthly Electricity Consumption"
    )

    # Bar chart
    bill_fig = px.bar(
        filtered,
        x="Month",
        y="Electricity_Bill",
        color="City",
        title="Monthly Electricity Bill"
    )

   # Calculate appliance-wise units
    appliance_data = filtered.groupby(
        "Appliance", as_index=False
    )["Units_Consumed"].sum()

 # Create pie chart
    appliance_fig = px.pie(
        appliance_data,
        names="Appliance",
        values="Units_Consumed",
        title="Electricity Consumption by Appliance"
    )
  # Return all dashboard outputs
    return summary, units_fig, bill_fig, appliance_fig
```

Step 7: Run the dashboard

```python
# Run dashboard
app.run(
    jupyter_mode="inline",
    debug=False,
    port=8050
)
```
