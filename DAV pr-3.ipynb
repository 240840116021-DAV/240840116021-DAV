Aim:- Apply ƒiltering, sorting, aggregation, and GroupBy operations on real-world datasets.

```python
# Import the pandas library for working with data
import pandas as pd

# Create a dictionary containing employee information
data= {
    "Emp_ID": [101, 102, 103, 104, 105, 106, 107, 108],   # Employee IDs
    "Name": ["Rahul", "Priya", "Amit", "Riya", "Karan", "Neha", "Vivek", "Pooja"],  # Employee names
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance", "IT", "HR"],    # Departments
    "City": ["Surat", "Navsar", "Surat", "Valsad", "Surat", "Navsari", "Valsad", "Surat"],     # Employee cities
    "Salary": [45000, 38000, 52000, 48000, 40000, 51000, 60000, 42000],    # Years of experience
    "Experience": [2, 1, 5, 4, 2, 6, 8, 3]
}
```

Step 1- Display the Dataset

```python
# Convert the dictionary into a pandas DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
df
```

Step 2- Filter Data
(Display employees from IT department)

```python
#Select only emplyoees working in the IT department.
#The condition returns rows where department is equal tp "IT"
df[df["Department"]=="IT"]
```

Display Employees with Salary Greater than 50000

```python
#Display Employees with Salary Greater than 50000
df[df["Salary"]>50000]
```

Display Employees from Surat

```python
# Filter rows where the City is "Surat"
df[df["City"]=="Surat"]
```

Apply Multiple Conditions

```python
# Filter rows where Department is "IT" and Salary is greater than 50,000
df[(df['Department']=="IT") & (df['Salary']>50000)]
```

Step 3- Sort Data

Sort salary in asecending order

```python
# Sort employees by salary from lowest to highest
df.sort_values(by='Salary')
```

Sort Salary in Descending Order

```python
# Sort the DataFrame by Salary in descending order
df.sort_values(by="Salary",ascending=False)
```

Sort by Department and Salary

```python
#first sort by department
# within each department sort salaries in descending order
df.sort_values(by=["Department","Salary"],ascending=[True,False])
```

Step 4 - Perform Aggregration

Calculate the salary of all Employees

```python
# Calculate the total salary
df["Salary"].sum()
```

Calculate Average Salary

```python
# Calculate the average salary
df["Salary"].mean()
```

Find Highest salary

```python
# Find the maximum salary
df["Salary"].max()
```

Find Lowest salary

```python
# Find the minimum salary
df["Salary"].min()
```

Count Employees

```python
# Count the number of employee IDs
df["Emp_ID"].count()
```

Step 5- Perform GroupBy Operations

Average Salary Department-wise

```python
# Group employees based on Department.
# Calculate the average salary of each department.
df.groupby("Department") ["Salary"].mean()
```

Minimum salary Department-wise

```python
# Find the minimum salary in each department
df.groupby("Department") ["Salary"].min()
```

Count the total employees department-wise

```python
# Count the number of employees in each department
df.groupby("Department")["Emp_ID"].count()
```

Total salary Department-wise

```python
# Calculate the total salary for each department
df.groupby("Department")["Salary"].sum()
```

Maximum Salary department-wise

```python
# Find the maximum salary in each department
df.groupby("Department") ["Salary"].max()
```

Group by city

```python
#Calculate the average salary of employees in each city
df.groupby("City") ["Salary"].mean()
```

Step 6- Multiple Aggregration Funtions

```python
# Calculate count, total, average, minimum, and maximum salary for each department
df.groupby("Department")["Salary"].agg(["count", "sum", "mean", "min", "max"])
```

Step 7- Save the Result

```python
#Save the dataset into CSV file.
df.to_csv("Employee_Data.csv", index=False)
print("Dataset saved successfully.")
```
