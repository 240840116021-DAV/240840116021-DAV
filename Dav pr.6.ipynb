# **Step 1- Create a Dataset**

```python
import pandas as pd
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Month":["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Product":["Laptop", "Mobile", "Tablet","Laptop", "Mobile", "Tablet","Laptop", "Mobile", "Tablet","Laptop", "Mobile", "Tablet"],
    "Region":["North", "South", "East", "West","North", "South", "East", "West","North", "South", "East", "West"],
    "Sales":[120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000, 250000],
    "Units_sold":[12, 30, 25, 14, 35, 30, 18, 38, 32, 21, 45, 40]
}
```

```python
df=pd.DataFrame(data)
print(df)
```

# **Task 2- Create a line chart**

How did the company's sales change from January to December?

```python
plt.figure(figsize=(10,5))

plt.plot(df["Month"], df["Sales"], marker="o")

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Company Sales from January to December")
plt.grid()
plt.show()
```

**Interpretation:-** The graph shows that sales increased month by month.
Sales were lowest in January and highest in December.
So, the company’s sales improved throughout the year.

# **Task-3 Create a Bar Chart**

Which Product generated how much total sales?

```python
plt.figure(figsize=(7,5))

plt.bar(df["Product"], df["Sales"], color="skyblue")


plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()
```

**Interpretation:-** The bar chart shows the sales of different products.
Tablet has the highest sales, while Laptop has the lowest sales.
Overall, Tablet is the best-selling product

# **Task 4- Create a Histogram**

How are the individual sales values distributed?

```python
plt.hist(df["Sales"], bins=5, edgecolor="black")

plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.title("Distribution of Individual Sales Values")
plt.show()
```

 **Interpretation**:- The histogram shows that most sales values are in the higher range.West region has the highest sales, while North region has the lowest sales.The higher sales values are mainly from the West and East regions.
Overall the sales values show a higher distribution.

# **Task 5- Create a Scatter Plot**

Is there a relationship between the number of unirs sold and total sales?

```python
plt.scatter(df["Units_sold"], df["Sales"])

plt.xlabel("Units Sold")
plt.ylabel("Total Sales")
plt.title("Relationship Between Units Sold and Total Sales")
plt.grid()
plt.show()
```

**Interpretation:-** The scatter plot shows a positive relationship between units sold and sales.When more units are sold, sales are generally higher.
So, higher units sold lead to higher sales.

# **Task 6- Create a Pie Chart**

What percentage of total sales comes from each region?

```python
region_sales = df.groupby("Region")["Sales"].sum()

# Create pie chart
plt.pie(region_sales, labels=region_sales.index, autopct="%1.1f%%")

plt.title("Sales Percentage by Region")
plt.show()
```

**Interpretation:-** The pie chart shows the percentage of total sales from each region.

-West has the highest sales: 26.7%

-East has 26.5%

-South has 24.6%

-North has the lowest: 22.2%

 Overall, West region contributes the most to total sales.
