Step 1

```python
import pandas as pd
```

Step 2 Create a sample data

```python
data = {
    "ID": [101, 102, 103, 104, 105, 106, 107],  # List of student IDs
    "NAME": ["Rahul", "Priya", "Amit", "Riya", "Karan", "Krisha", "Harry"],  # List of student names
    "Age": [20, 21, 19, 20, 22, 24, 28],    # List of student ages
    "City": ["Surat", "Navsari", "Valsad", "Bharuch", "Surat", "Gandhinagar", "Ahemdabad"],  # List of cities where students live
    "Marks": [85, 92, 78, 88, 81, 84, 90] # List of student marks
}
```

Step 3 Create a DataFrame

```python
df = pd.DataFrame(data) # Create a Pandas DataFrame from the dictionary
df # Display the DataFrame
```

Step 4 Save Data as a CSV file

```python
#Save the DataFrame in CSV file named as "students.csv"
#index=False prevents the DataFrame index from being saved as a column
df.to_csv("students.csv", index=False)

# Display a confirmation message
print("CSV file created successfully.")
```

Step 5 Save Data as a Excel file

```python
# Save DataFrame to Excel
df.to_excel("students.xlsx", index=False)

# Display a confirmation message
print("Excel file created successfully.")
```

Step 6 Save Data as a JSON file

```python
# Save DataFrame to JSON
df.to_json("students.json", orient="records", indent=4)

# Display a confirmation message
print("JSON File created successfully.")
```
